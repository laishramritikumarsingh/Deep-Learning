#7. Loss Function Implementation

1. Import Libraries

```python
import numpy as np
```

2. Define actual and predicted values

```python
# Actual class labels
y_actual = np.array([1, 0, 1, 1, 0])

# Predicted probabilities
y_predicted = np.array([0.9, 0.2, 0.8, 0.6, 0.1])

print("Actual Values:   ", y_actual)
print("Predicted Values:", y_predicted)
```

3. Calculate Mean Squared Error

```python
mse = np.mean((y_actual - y_predicted) ** 2)

print(f"Mean Squared Error: {mse:.4f}")
```

4. Calculate Cross-Entropy-Loss

```python
# Avoid log(0)
epsilon = 1e-15

y_predicted_clipped = np.clip(y_predicted, epsilon, 1 - epsilon)

cross_entropy = -np.mean(
    y_actual * np.log(y_predicted_clipped) +
    (1 - y_actual) * np.log(1 - y_predicted_clipped)
)

print("Cross-Entropy Loss:", cross_entropy)
```

5. Loss Function Comparison

```python
print("Loss Function Comparison")
print("------------------------")
print(f"Mean Squared Error : {mse:.4f}")
print(f"Cross-Entropy Loss : {cross_entropy:.4f}")
```
